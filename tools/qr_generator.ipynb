```python
from pathlib import Path

import qrcode
from IPython.display import display
```

```python
# Cambia este valor por la URL o texto que quieras codificar.
contenido = "marinaserantescasals.azurewebsites.net"
nombre_archivo = "codigo-qr.png"

qr = qrcode.QRCode(
    version=None,
    error_correction=qrcode.constants.ERROR_CORRECT_M,
    box_size=12,
    border=4,
)
qr.add_data(contenido)
qr.make(fit=True)
imagen = qr.make_image(fill_color="#292d26", back_color="#eeeae2")

ruta_salida = Path(nombre_archivo)
imagen.save(ruta_salida)
display(imagen)
print(f"QR guardado en: {ruta_salida.resolve()}")
```
